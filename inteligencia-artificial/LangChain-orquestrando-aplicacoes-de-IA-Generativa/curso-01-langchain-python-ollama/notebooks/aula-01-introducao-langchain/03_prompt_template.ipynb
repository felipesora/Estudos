# 03 — Prompt Template com LangChain

Neste notebook vamos evoluir o exemplo anterior utilizando **templates de prompt**.

Até agora tínhamos um prompt fixo:

```python
prompt = "Explique o que é a inteligência artificial em poucas palavras"
```

O problema é que, se quisermos fazer a mesma pergunta sobre vários conceitos, precisamos alterar a string manualmente.

Com `PromptTemplate`, podemos criar um modelo reutilizável e substituir apenas as informações que mudam.

### Fluxo

```text
Template
   ↓
Variáveis
   ↓
Prompt final
   ↓
Qwen
```

## 1. Importando o PromptTemplate

Além do `ChatOllama`, agora vamos importar `PromptTemplate`.

`PromptTemplate` é um componente do LangChain usado para criar prompts reutilizáveis.

````python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
````

## 2. Configurando a LLM

Continuamos utilizando o Qwen através do Ollama local.

````python
llm = ChatOllama(
    model="qwen3:4b",
    base_url="http://localhost:11434"
)
````

## 3. Criando um template

Agora vamos criar nosso primeiro template usando `from_template()`.

A parte entre `{}` representa uma **variável** que poderá ser preenchida posteriormente.

Neste caso, `{conceito}` é a variável.

````python
prompt_template = PromptTemplate.from_template(
    "Explique o conceito de {conceito} em poucas palavras."
)
````

### O que aconteceu aqui?

Criamos um template que pode ser representado assim:

```text
Explique o conceito de {conceito} em poucas palavras.
```

`{conceito}` ainda não possui um valor. Ele será preenchido quando utilizarmos o template.

## 4. Preenchendo a variável

Agora vamos fornecer o valor da variável `conceito`.

Usamos `invoke()` no `PromptTemplate` e passamos um dicionário contendo o nome da variável e seu valor.

````python
prompt = prompt_template.invoke({
    "conceito": "inteligência artificial"
})
````

O template agora foi preenchido.

O resultado representa o prompt final:

```text
Explique o conceito de inteligência artificial em poucas palavras.
```

````python
print(prompt)
````

## 5. Enviando o prompt para o Qwen

Agora que temos o prompt preenchido, podemos enviá-lo para a LLM usando `llm.invoke()`.

````python
response = llm.invoke(prompt)
````

## 6. Exibindo a resposta

````python
print(response.content)
````

## 7. Por que usar templates?

A principal vantagem é a reutilização.

Podemos utilizar o mesmo template para vários conceitos sem alterar a estrutura do prompt.

````python
conceitos = ["Python", "machine learning", "RAG", "LangChain"]

for conceito in conceitos:
    prompt = prompt_template.invoke({
        "conceito": conceito
    })

    response = llm.invoke(prompt)

    print(f"\n--- {conceito} ---")
    print(response.content)
````

## 8. Código completo

Depois de entender cada etapa, podemos juntar tudo:

````python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate

llm = ChatOllama(
    model="qwen3:4b",
    base_url="http://localhost:11434"
)

prompt_template = PromptTemplate.from_template(
    "Explique o conceito de {conceito} em poucas palavras."
)

prompt = prompt_template.invoke({
    "conceito": "inteligência artificial"
})

response = llm.invoke(prompt)

print(response.content)
````
