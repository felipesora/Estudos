# 02 — Demonstração com LangChain

Neste notebook vamos fazer a mesma tarefa do notebook anterior, mas agora utilizando o **LangChain**.

No notebook `01_demonstracao_sem_langchain`, usamos diretamente o cliente do Ollama:

```text
Python → Ollama → Qwen
```

Agora vamos adicionar o LangChain entre nossa aplicação e o Ollama:

```text
Python → LangChain → Ollama → Qwen
```

O objetivo é entender o que o LangChain está abstraindo e como podemos conversar com uma LLM através dele.

## 1. Importando o ChatOllama

Vamos importar `ChatOllama` do pacote `langchain_ollama`.

`ChatOllama` é a integração do LangChain com o Ollama e permite utilizar modelos que estão sendo executados localmente.

````python
from langchain_ollama import ChatOllama
````

## 2. Criando o objeto da LLM

Agora criamos um objeto `ChatOllama`.

Esse objeto representa nossa LLM dentro do LangChain.

````python
llm = ChatOllama(
    model="qwen3:4b",
    base_url="http://localhost:11434"
)
````

### O parâmetro `model`

```python
model="qwen3:4b"
```

Aqui informamos qual modelo do Ollama queremos utilizar.

Neste projeto estamos utilizando o **Qwen 3 de 4 bilhões de parâmetros**, que foi baixado anteriormente pelo Ollama.

### O parâmetro `base_url`

```python
base_url="http://localhost:11434"
```

Esse endereço informa ao LangChain onde encontrar o servidor Ollama.

Como estamos executando o Ollama localmente através do Docker, usamos a porta padrão `11434`.

## 3. Criando o prompt

Assim como no exemplo anterior, criamos uma variável contendo a pergunta que será enviada para a LLM.

````python
prompt = "Explique o que é a inteligência artificial em poucas palavras"
````

## 4. Enviando o prompt para a LLM

Agora utilizamos o método `invoke()` do LangChain:

````python
response = llm.invoke(prompt)
````

### O que é `invoke()`?

O método `invoke()` é uma das formas padronizadas do LangChain de executar um componente.

Neste caso:

```text
prompt
  ↓
llm.invoke()
  ↓
ChatOllama
  ↓
Ollama
  ↓
Qwen
```

O LangChain recebe nosso prompt, utiliza a integração `ChatOllama` e envia a solicitação para o modelo que está rodando no Ollama.

## 5. Visualizando a resposta

Diferentemente do cliente direto do Ollama, o LangChain retorna um objeto de mensagem.

O conteúdo textual da resposta está disponível através da propriedade `.content`.

````python
print(response.content)
````

## 6. Inspecionando a resposta

Como `response` é um objeto do LangChain, podemos observar seu tipo:

````python
print(type(response))
````

Também podemos acessar somente o conteúdo textual:

````python
response.content
````

## 7. Código completo

Depois de entender cada parte, podemos juntar tudo:

````python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="qwen3:4b",
    base_url="http://localhost:11434"
)

prompt = "Explique o que é a inteligência artificial em poucas palavras"

response = llm.invoke(prompt)

print(response.content)
````

## 8. Comparando com o exemplo sem LangChain

No exemplo anterior utilizamos diretamente o cliente do Ollama:

```python
from ollama import Client

client = Client(host="http://localhost:11434")

response = client.chat(...)
```

Agora utilizamos a abstração do LangChain:

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(...)

response = llm.invoke(prompt)
```

A ideia central do LangChain é fornecer uma interface mais padronizada para trabalhar com modelos e construir aplicações mais complexas em cima deles.
