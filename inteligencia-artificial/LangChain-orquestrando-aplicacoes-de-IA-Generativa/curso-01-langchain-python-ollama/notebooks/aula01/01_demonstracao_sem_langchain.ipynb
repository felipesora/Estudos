# 01 — Demonstração sem LangChain

Neste notebook vamos reproduzir o primeiro exemplo do projeto **sem usar LangChain**.

A ideia é entender o funcionamento básico antes de adicionar abstrações como LangChain.

### Arquitetura

```text
Python → Ollama → Qwen
```

O Python envia um `prompt` para o Ollama, que executa o modelo **Qwen** localmente e devolve a resposta.

## 1. Importando o cliente do Ollama

Primeiro importamos `Client` da biblioteca `ollama`.

Esse objeto será responsável por permitir que nosso código Python se comunique com o servidor Ollama.

````python
from ollama import Client
````

## 2. Conectando ao Ollama

Agora criamos um cliente apontando para o endereço onde o Ollama está rodando:

```text
http://localhost:11434
```

- `localhost` significa que o serviço está sendo executado na nossa própria máquina.
- `11434` é a porta padrão da API do Ollama.

Neste projeto, o Ollama está sendo executado em um container Docker, mas a porta `11434` é disponibilizada para o computador.

````python
client = Client(host="http://localhost:11434")
````

## 3. Criando o prompt

Um **prompt** é a entrada que enviamos para a LLM.

Neste primeiro exemplo, vamos fazer uma pergunta simples.

````python
prompt = "Explique o que é a inteligência artificial em poucas palavras"
````

## 4. Enviando o prompt para o modelo

Agora utilizamos o método `chat()` do cliente Ollama.

Passamos duas informações principais:

- `model`: informa qual modelo deve responder.
- `messages`: contém as mensagens da conversa.

O papel (`role`) `user` indica que aquela mensagem foi enviada pelo usuário.

````python
response = client.chat(
    model="qwen3:4b",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ]
)
````

## 5. Visualizando a resposta

O Ollama retorna um objeto contendo informações sobre a resposta.

A resposta textual do modelo fica dentro de:

```text
response["message"]["content"]
```

Vamos imprimir somente o texto gerado pela LLM.

````python
print(response["message"]["content"])
````

## 6. Código completo

Depois de entender cada parte, podemos juntar tudo:

````python
from ollama import Client

client = Client(host="http://localhost:11434")

prompt = "Explique o que é a inteligência artificial em poucas palavras"

response = client.chat(
    model="qwen3:4b",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ]
)

print(response["message"]["content"])
````
